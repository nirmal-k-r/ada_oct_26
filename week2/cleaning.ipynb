```python
df.rename(columns={'NA_Sales': "Na_sales", 'EU_Sales': "Eu_sales", 'JP_Sales': "Jp_sales", 'Other_Sales': "Other_sales", 'Global_Sales': "Global_sales"}, inplace=True)
df.columns
```

```python
# Preprocessing
# Identify null values in the dataset

df.isnull().sum()
```

```python
df2=df.dropna()
# df.dropna(inplace=True)
df2.head(10)
```

```python
df.head(10)
```

```python
df3=df.copy()

df3['Platform'].fillna('Not available', inplace=True)
df3.head(10)
```

```python
# Check for duplicates in the dataset
df.duplicated().sum()
```

```python
#remove duplicates
df.drop_duplicates(inplace=True)

df.duplicated().sum()
```

```python
# GHandling extreme values in the dataset

#visualise some of the columns to check for extreme values

df['Global_sales'].plot(kind='box', title='Global Sales', figsize=(10, 5))
```

```python
#filter eveyrthing higher than 10
df4=df[df['Global_sales']<10]

df4['Global_sales'].plot(kind='box', title='Global Sales', figsize=(10, 5))
```

```python
#filter eveyrthing higher than 10
df5=df[df['Global_sales']<0.7]

df5['Global_sales'].plot(kind='box', title='Global Sales', figsize=(10, 5))
```

```python
df5
```

```python
# Encode action column
# import LabelEncoder
from sklearn.preprocessing import LabelEncoder

genre_le=LabelEncoder()
genre_le.fit(df['Genre'])
df5['Genre_encoded']=genre_le.transform(df5['Genre'])

df5.head(10)
```

```python
df5.describe()
```

```python
#normalisation (minxmax or standard_scaler)
#normalise total sales columns
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df5['Global_sales_normalised'] = scaler.fit_transform(df5[['Global_sales']])
df5.head(10)


```

```python
df5['EUUS_sales'] = df5['Eu_sales'] + df5['Na_sales']
df5
```

```python
df5['Platform'].unique()
```

```python
gaming_platforms=df5['Platform'].value_counts()
gaming_platforms
```

```python
top_5_gaming_platforms=df5['Platform'].value_counts().head(5)
top_5_gaming_platforms
```

```python
top_5_gaming_platforms.plot(kind='bar', title='Top Gaming Platforms', figsize=(10, 5), color=['blue', 'orange', 'green', 'red', 'purple'])
plt.show()
```

```python
top_5_gaming_platforms.plot(kind='pie', title='Top 5 Gaming Platforms', figsize=(10, 5))
```

```python
genre_popularity.plot(kind='bar', title='Genre Popularity', figsize=(10, 5), color=['blue', 'orange', 'green', 'red', 'purple'])
```

```python
# Regional video game sales analysis
df6=df5[['Eu_sales', 'Na_sales', 'Jp_sales', 'Other_sales']] # Select the sales columns

comparison=df6.sum() # summed all their sales

comparison.plot(kind='bar', title='Regional Video Game Sales Comparison', figsize=(10, 5), color=['blue', 'orange', 'green', 'red'])


```

```python
df5.groupby('Platform')['Global_sales'].sum().sort_values(ascending=False).head(10).plot(kind='bar', title='Top 10 Gaming Platforms by Global Sales', figsize=(10, 5), color=['blue', 'orange', 'green', 'red', 'purple'])
```
