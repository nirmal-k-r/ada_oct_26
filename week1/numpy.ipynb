```python
#why need to use numpy array instead of list?
marks=[150,70,89,98,29,100,67]

total=0
for mark in marks:
    total+=mark
print(total)
average=total/len(marks)
print(average)
```

```python
marks=[150,70,89,98,29,100,67]

marks_array=np.array(marks)
print(marks_array)

print(np.average(marks_array))

print(np.sum(marks_array))

print(np.max(marks_array))

print(np.min(marks_array))


print(np.median(marks_array))

print(np.std(marks_array))
```

```python
x=[1,2,3,4,5]
y=[6,7,8,9,10]

z=[]

for i, val in enumerate(x):
    x_val=x[i]
    y_val=y[i]
    z.append(x_val+y_val)
print(z)
```

```python
x=[1,2,3,4,5]
y=[6,7,8,9,10]

x_array=np.array(x)
y_array=np.array(y)

z_array=x_array+y_array
print(z_array)
```

```python
sub=x_array-y_array
print(sub)

multiplication=x_array*y_array
print(multiplication)

div=x_array/y_array
print(div)

rounded=np.round(div,2)
print(rounded)

```
